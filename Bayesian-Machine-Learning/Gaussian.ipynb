# Gaussian Processes I — Theory Summary

This notebook summarizes the **theory** from the *Gaussian Processes I* slides, focusing on core ideas and formulas rather than worked examples. Headings are descriptive only (no manual numbering).

## Function–space view (informal definition)

A **Gaussian process (GP)** is a distribution over functions $f:\mathcal{X}\to\mathbb{R}$ such that any finite collection $(f(x_1),\ldots,f(x_n))$ is jointly Gaussian:
$$
f \sim \mathcal{GP}\big(m(x),\,k(x,x')\big),\qquad m(x)=\mathbb{E}[f(x)],\; k(x,x')=\mathrm{cov}\!\left[f(x),f(x')\right].
$$

In practice, the choice of **kernel** encodes our prior assumptions (smoothness, periodicity, etc.) and determines how information is shared across inputs. The posterior combines the prior with observed data to yield predictions with uncertainty bands (see page 14 figure).

## Review: multivariate Gaussian distribution

For $x\in\mathbb{R}^D$, the density is
$$
p(x\mid \mu,\Sigma)=(2\pi)^{-D/2}\,|\Sigma|^{-1/2}\,\exp\!\left(-\tfrac12 (x-\mu)^\top \Sigma^{-1}(x-\mu)\right),
$$
with mean $\mu\in\mathbb{R}^D$ and **positive-definite** covariance $\Sigma\in\mathbb{R}^{D\times D}$.  
Linear combinations $a^\top x$ are Gaussian for any nonzero $a\in\mathbb{R}^D$ (definition slide).

### Interpreting the covariance matrix

- Diagonal entries control **marginal variances**; off-diagonals set **covariances** $(\Sigma)_{ij}=\mathrm{cov}(x_i,x_j)$.  
- Valid covariance matrices are **symmetric** and **positive definite**; constraints like $|c|\le \sqrt{ab}$ appear for $2\times2$ cases with $\Sigma=\begin{bmatrix}a&c\\c&b\end{bmatrix}$.  
- The **mode density** at $x=\mu$ scales with $|\Sigma|^{-1/2}$ (smaller determinant → sharper peak).  
See pages 24–29 for illustrations and checks of valid covariances.

### Useful Gaussian identities

- **Closure under addition:** if $x_1\sim\mathcal{N}(m_1,V_1)$ and $x_2\sim\mathcal{N}(m_2,V_2)$ (independent), then $x_1+x_2\sim\mathcal{N}(m_1+m_2,\,V_1+V_2)$.
- **Affine transformation:** $Ax+b\sim\mathcal{N}(Am+b,\,AV A^\top)$.  
(See page 30.)

### Marginalization and conditioning

Partition $x=\begin{bmatrix}x_1\\x_2\end{bmatrix}\sim\mathcal{N}\!\left(\!\begin{bmatrix}m_1\\m_2\end{bmatrix},\begin{bmatrix}\Sigma_{11}&\Sigma_{12}\\\Sigma_{21}&\Sigma_{22}\end{bmatrix}\!\right)$. Then
- **Marginals:** $x_1\sim\mathcal{N}(m_1,\Sigma_{11})$, $x_2\sim\mathcal{N}(m_2,\Sigma_{22})$.
- **Conditional:** 
$$
x_1\mid x_2 \sim \mathcal{N}\!\Big(m_1+\Sigma_{12}\Sigma_{22}^{-1}(x_2-m_2),\;\Sigma_{11}-\Sigma_{12}\Sigma_{22}^{-1}\Sigma_{21}\Big).
$$
Pages 31–36 show 2D examples of how the conditional mean shifts and the variance contracts.

## Bayesian linear regression (weight-space view)

Assume $y_i = w^\top x_i + \varepsilon_i$ with $\varepsilon_i\sim\mathcal{N}(0,\sigma_{\text{obs}}^2)$. The likelihoods are
$$
p(y_i\mid x_i,w)=\mathcal{N}(y_i\mid w^\top x_i,\sigma_{\text{obs}}^2),\qquad p(y\mid X,w)=\mathcal{N}(y\mid Xw,\sigma_{\text{obs}}^2 I).
$$
Choose a **Gaussian prior** $p(w)=\mathcal{N}(0,\Sigma_p)$.  
Then the **posterior** is Gaussian
$$
p(w\mid y)=\mathcal{N}\!\big(\mu,\;A^{-1}\big),\qquad A=\tfrac{1}{\sigma_{\text{obs}}^2}X^\top X+\Sigma_p^{-1},\quad \mu=\tfrac{1}{\sigma_{\text{obs}}^2}A^{-1}X^\top y.
$$
(See pages 41–44.)

### Predictive and uncertainty decomposition

For a new input $x_{*}$, the **predictive distribution** for $y_{*}$ (including observation noise) is
$$
p(y_{*}\mid y)=\mathcal{N}\!\big(y_{*}\mid \mu^\top x_{*},\; \sigma_{\text{obs}}^2 + x_{*}^\top A^{-1} x_{*}\big).
$$
- $\sigma_{\text{obs}}^2$: measurement noise,  
- $x_{*}^\top A^{-1} x_{*}$: **epistemic** uncertainty from limited data (posterior over $w$).
See pages 53–54.

## From linear regression to Gaussian processes

Bayesian linear regression with features $\phi(x)$ induces the kernel
$$
k(x,x')=\phi(x)^\top \Sigma_p \phi(x'),
$$
and the corresponding function values are jointly Gaussian. Moving to a GP replaces explicit features by a **kernel** that directly specifies prior correlations across inputs; the posterior then yields smooth predictive means and credible intervals as shown in the *GP paradigm* figure (page 14).
