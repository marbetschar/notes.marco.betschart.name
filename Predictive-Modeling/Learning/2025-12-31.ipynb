## **SOLUTION SKETCH**

### **Part A Solutions:**

**A1(a) - Test Statistic:**
$t^* = \frac{-5.344 - 0}{0.559} = -9.560$

- df = 30; Critical value: $\pm 2.042$
- **Decision:** Reject $H_0$ (|-9.560| > 2.042)
- **Conclusion:** Vehicle weight has a statistically significant effect on MPG (p < 0.001)

**A1(b) - Confidence Interval:**
$-5.344 \pm 2.042 \times 0.559 = -5.344 \pm 1.141 = [-6.485, -4.203]$

- **Interpretation:** Each additional thousand pounds of weight is associated with a 4.2 to 6.5 MPG decrease on average

**A1(c) - Predictions:**

- $\hat{y}_0 = 37.285 - 5.344(3) = 21.252$ MPG
- Confidence Interval (E[Y|x₀]): Approximately [20.1, 22.4] MPG
- Prediction Interval (Y|x₀): Approximately [14.7, 27.8] MPG
- The PI is wider because it must account for both estimation uncertainty and individual observation variability

**A2(a) - Violation:** Heteroscedasticity (non-constant variance) of error terms. The assumption $\text{Var}(\varepsilon_i) = \sigma^2$ is violated; actually $\text{Var}(\varepsilon_i) = \sigma_i^2$ where $\sigma_i$ increases with fitted value.

**A2(b) - Consequences:**

1. Coefficients remain **unbiased** but inefficient
2. Standard errors are **downward-biased** (underestimated)
3. Confidence intervals are **too narrow** (nominal level < true coverage)
4. p-values are **too small**; Type I error inflation
5. The test statistic is still valid, but reported p-values are misleading

**A2(c) - Remedies:**

1. **Log-transformation:** $Y^* = \log(\text{MPG})$ often stabilizes variance, especially for rate data
2. **WLS:** Weight observations inversely by estimated variance; $\min \sum_i w_i(y_i - \hat{y}_i)^2$ where $w_i = 1/\hat{\sigma}_i^2$
